1683. Invalid Tweets

Write a solution to find the IDs of the invalid tweets. 
The tweet is invalid if the number of characters used in the content of the tweet is strictly greater than 15.

Return the result table in any order.

```python
from pyspark.sql import SparkSession as ss
from pyspark.sql.functions import *
import pandas as pd
import numpy as np
```

```python
spark = (
    ss.builder.appName('1683. Invalid Tweets').master('local[*]').getOrCreate()
)
spark 
```

```python
_data = [[1, 'Let us Code'], [2, 'More than fifteen chars are here']]
_schema = "tweet_id int, content string"

df_tweets = spark.createDataFrame(data=_data, schema=_schema)
df_tweets.show(truncate=False)
```

```python
df_tweets.select('tweet_id').filter(len('content') > 15).show()
```
