A country is big if:

it has an area of at least three million (i.e., 3000000 km2), or
it has a population of at least twenty-five million (i.e., 25000000).
Write a solution to find the name, population, and area of the big countries.

```python
import pandas as pd
import numpy as np
from pyspark.sql import SparkSession as ss
from pyspark.sql.functions import *
```

```python
spark = (
    ss\
    .builder\
    .appName('Leetcode: 595')\
    .master('local[*]')\
    .getOrCreate()
)

spark
```

```python
_data = [
    ['Afghanistan','Asia',652230,25500100,20343000000],
    ['Albania','Europe',28748,2831741,12960000000],
    ['Aglgeria','Africa',2381741,37100000,188681000000],
    ['Andorra','Europe',468,78115,3712000000],
    ['Angola','Africa',1246700,20609294,100990000000]
]

_schema = "name string, continent string, area long, population long, gdp long"
```

```python
df_world = spark.createDataFrame(data=_data, schema=_schema)

df_world.printSchema()
```

```python
df_world\
.select('name','population','area')\
.filter((col('area') >= 3000000) | (col('population') >= 25000000))\
.show()
```
