```python
import pandas as pd
import numpy as np
from pyspark.sql import SparkSession as ss
from pyspark.sql.functions import *
```

```python
spark = (
    ss\
    .builder\
    .appName('Leetcode 584')\
    .master('local[*]')\
    .getOrCreate()
)
```

```python
_data = [
    [1, 'Will', None],
    [2, 'Jane', None],
    [3, 'Alex', 2],
    [4, 'Bill', None],
    [5, 'Zack', 1],
    [6, 'Mark', 2]
]

_schema = "id int, name string, refree_id int"

df_customer = spark.createDataFrame(data=_data, schema=_schema)

df_customer.printSchema()
```

Find the names of the customer that are either:

referred by any customer with id != 2.
not referred by any customer.
Return the result table in any order.

```python
df_customer\
.select('name')\
.filter((col('refree_id') != 2) | isnull(col('refree_id')))\
.show()
```
