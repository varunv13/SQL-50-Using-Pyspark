1378. Replace Employee ID With The Unique Identifier

Write a solution to show the unique ID of each user, If a user does not have a unique ID replace just show null.
Return the result table in any order.

```python
from pyspark.sql import SparkSession as ss
from pyspark.sql.functions import *
import pandas as pd
import numpy as np
```

```python
spark = ss.builder.appName('1378. Replace Employee ID With The Unique Identifier').master('local[*]').getOrCreate()
```

```python
_data_emp = [[1,'Alice'],[7,'Bob'],[11,'Meir'],[90,'Winston'],[3,'Jonathan']]
_data_empuni = [[3,1],[11,2],[90,3]]

_schema_emp = "id int, name string"
_schema_empuni = "id int, unique_id string"

df_employees   = spark.createDataFrame(data=_data_emp, schema=_schema_emp)
df_employeeUni = spark.createDataFrame(data=_data_empuni, schema=_schema_empuni)

df_employees.show()
df_employeeUni.show()
```

```python
df_joined = df_employees.join(df_employeeUni, df_employees.id == df_employeeUni.id ,'left')
df_joined.show()
```

```python
df_joined.select('unique_id','name').show()
```
