```python
import pandas as pd
import numpy as np
from pyspark.sql import SparkSession as ss
from pyspark.sql.functions import *
```

```python
spark = ss.builder.appName('Leetcode Question: 1757').master('local[*]').getOrCreate() 
        # .builder                                          # builder is used to provide configuration to a spark application
        # .appName('Leetcode Question: 1757')               # appName() provides a name to your spark application  
        # .master('local[*]')                               # master() tells where spark shougld run and how many cores it shogld utlize 
        # .getOrCreate()                                    # getOrCreate() tells that if a spark session already exists then give me that one 
```

```python
_data = [
    [0, 'Y', 'N'],
    [1, 'Y', 'Y'],
    [2, 'N', 'Y'],
    [3, 'Y', 'Y'],
    [4, 'N', 'N']
]

_col = "product_id int, low_fats string, recyclable string"

df_products = spark.createDataFrame(data = _data, schema = _col);
```

```python
df_products.printSchema()
df_products.show()
```

Write a solution to find the ids of products that are both low fat and recyclable.
Return the result table in any order.

```python
df_products\
.select('product_id')\
.filter((col('low_fats') == 'Y') & (col('recyclable') == 'Y'))\
.orderBy(col('product_id'))\
.show()
```
